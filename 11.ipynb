```python
import numpy as np
import torch 
from torch import nn
import matplotlib.pyplot as plt

m = 0.1234
c = 0.6789

# Correct range
X = np.arange(0.04, 0.2, 0.01)
y = m * X + c

split = int(0.8 * len(X))
X_train, y_train = X[:split], y[:split]
X_test, y_test = X[split:], y[split:]

# Convert to tensors
X_train_tensors = torch.from_numpy(X_train.astype(np.float32))
y_train_tensors = torch.from_numpy(y_train.astype(np.float32))
X_test_tensors = torch.from_numpy(X_test.astype(np.float32))
y_test_tensors = torch.from_numpy(y_test.astype(np.float32))

print(f"Shape of X_train is {X_train.shape} and type is {type(X_train)}")
print(f"Shape of X_train_tensors is {X_train_tensors.shape} and type is {type(X_train_tensors)}")

def plot_data(X_train=X_train, y_train=y_train, X_test=X_test, y_test=y_test, predictions=None):
    plt.scatter(X_train, y_train, c='r', label="Training Data")
    plt.scatter(X_test, y_test, c='b', label="Testing Data")
    if predictions is not None:
        plt.scatter(X_test, predictions, c='g', label='Predicted Data')
    plt.legend()
    plt.show()

plot_data()


```
